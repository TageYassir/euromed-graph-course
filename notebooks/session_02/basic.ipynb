```python
import networkx as nx

S = nx.MultiDiGraph(name="Community")

people = ["Alice", "Bob", "Carol", "David", "Eve"]
S.add_nodes_from(people)

relationships = [
    ("Alice","Bob"),
    ("Bob","Carol"),
    ("Alice","David"),
    ("carol","Eve"),
    ("David","Eve")]

S.add_edges_from(relationships)
```

```python
def get_AllNodes(graph):
    count = 0
    while count < len(graph.nodes) - 1:
        count +=1
    return count
```

```python
get_AllNodes(S)
```

```python
def get_AllEdges(graph):
    count = 0
    while count < len(graph.edges):
        count +=1
    return count
```

```python
get_AllEdges(S)
```

```python
def find_neighbors(graph, node):
    neighbor = list(graph.neighbors(node))
    return neighbor
```

```python
find_neighbors(S, "Alice")
```

```python
def find_degree(graph, node):
    degree = graph.degree(node)
    return degree
```

```python
find_degree(S, "Alice")
```

```python
import matplotlib.pyplot as plt

def print_graph(graph):
   graph_visualization = nx.spring_layout(graph)
   nx.draw_networkx(graph, graph_visualization, with_labels=True, node_color='lightblue', edge_color='gray', node_size=2000, font_size=16, font_weight='bold')
   plt.show()
```

```python
print_graph(S)
```
